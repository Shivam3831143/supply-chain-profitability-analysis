```python
# CUSTOMER PROFITABILITY ANALYSIS

customer_analysis = df.groupby("Customer Id").agg(
    Total_Revenue=("Sales", "sum"),
    Total_Profit=("Benefit per order", "sum"),
    Total_Records=("Customer Id", "count"),
    Average_Profit=("Benefit per order", "mean"),
    Loss_Records=("Benefit per order", lambda x: (x < 0).sum())
).reset_index()

# Calculate customer profit margin
customer_analysis["Profit_Margin"] = (
    customer_analysis["Total_Profit"] /
    customer_analysis["Total_Revenue"]
) * 100

# Sort customers by total profit
top_profitable_customers = customer_analysis.sort_values(
    by="Total_Profit", ascending=False
)

# Customers with negative total profit
loss_making_customers = customer_analysis[
    customer_analysis["Total_Profit"] < 0
].sort_values(by="Total_Profit")

print("========== CUSTOMER ANALYSIS ==========")

print("\nTotal Unique Customers:",
      customer_analysis["Customer Id"].nunique())

print("\nTop 10 Profitable Customers:")
display(top_profitable_customers.head(10).round(2))

print("\nTop 10 Loss-Making Customers:")
display(loss_making_customers.head(10).round(2))

print("\nTotal Loss-Making Customers:",
      len(loss_making_customers))
```

```python
# CUSTOMER SEGMENT ANALYSIS

segment_analysis = df.groupby("Customer Segment").agg(
    Total_Revenue=("Sales", "sum"),
    Total_Profit=("Benefit per order", "sum"),
    Total_Records=("Customer Id", "count"),
    Unique_Customers=("Customer Id", "nunique"),
    Average_Profit=("Benefit per order", "mean")
).reset_index()

segment_analysis["Profit_Margin"] = (
    segment_analysis["Total_Profit"] /
    segment_analysis["Total_Revenue"]
) * 100

print("========== CUSTOMER SEGMENT PERFORMANCE ==========")
display(segment_analysis.round(2))


# LOSS-MAKING CUSTOMER PATTERNS

print("\n========== LOSS-MAKING CUSTOMER PATTERNS ==========")

one_record_losses = loss_making_customers[
    loss_making_customers["Total_Records"] == 1
]

multiple_record_losses = loss_making_customers[
    loss_making_customers["Total_Records"] > 1
]

print("Total Loss-Making Customers:",
      len(loss_making_customers))

print("Loss-Making Customers with One Record:",
      len(one_record_losses))

print("Loss-Making Customers with Multiple Records:",
      len(multiple_record_losses))

print("\nAverage Profit of One-Record Loss Customers:",
      round(one_record_losses["Total_Profit"].mean(), 2))

print("Average Profit of Multiple-Record Loss Customers:",
      round(multiple_record_losses["Total_Profit"].mean(), 2))
```

```python
# CUSTOMER CONTRIBUTION / PARETO ANALYSIS

customer_ranked = customer_analysis.sort_values(
    by="Total_Profit",
    ascending=False
).reset_index(drop=True)

total_customers = len(customer_ranked)
total_business_profit = customer_ranked["Total_Profit"].sum()

# Calculate cumulative profit contribution
customer_ranked["Cumulative_Profit"] = (
    customer_ranked["Total_Profit"].cumsum()
)

customer_ranked["Cumulative_Profit_Percentage"] = (
    customer_ranked["Cumulative_Profit"] /
    total_business_profit
) * 100

# Top 10% and top 20% customers
top_10_count = int(np.ceil(total_customers * 0.10))
top_20_count = int(np.ceil(total_customers * 0.20))

top_10_profit = customer_ranked.head(top_10_count)["Total_Profit"].sum()
top_20_profit = customer_ranked.head(top_20_count)["Total_Profit"].sum()

print("========== CUSTOMER CONTRIBUTION ANALYSIS ==========")

print("Total Customers:", total_customers)

print("\nTop 10% Customers:", top_10_count)
print("Profit Contribution:", round(top_10_profit, 2))
print("Share of Total Profit:",
      round((top_10_profit / total_business_profit) * 100, 2), "%")

print("\nTop 20% Customers:", top_20_count)
print("Profit Contribution:", round(top_20_profit, 2))
print("Share of Total Profit:",
      round((top_20_profit / total_business_profit) * 100, 2), "%")

print("\nTop 10 Customers:")
display(
    customer_ranked[
        ["Customer Id", "Total_Revenue", "Total_Profit",
         "Total_Records", "Cumulative_Profit_Percentage"]
    ].head(10).round(2)
)
```

```python
# CUSTOMER VALUE INDEX (CVI)

cvi = customer_analysis.copy()

# Convert each metric into percentile ranks
cvi["Profit_Score"] = cvi["Total_Profit"].rank(pct=True) * 100
cvi["Revenue_Score"] = cvi["Total_Revenue"].rank(pct=True) * 100
cvi["Frequency_Score"] = cvi["Total_Records"].rank(pct=True) * 100

# Weighted Customer Value Index
cvi["CVI"] = (
    0.50 * cvi["Profit_Score"] +
    0.30 * cvi["Revenue_Score"] +
    0.20 * cvi["Frequency_Score"]
)

# Classify customers
cvi["Customer_Value_Category"] = np.where(
    cvi["Total_Profit"] < 0,
    "Loss-making",
    np.where(
        cvi["CVI"] >= 80, "High Value",
        np.where(cvi["CVI"] >= 40, "Medium Value", "Low Value")
    )
)

# Summary
cvi_summary = cvi.groupby("Customer_Value_Category").agg(
    Customers=("Customer Id", "count"),
    Total_Revenue=("Total_Revenue", "sum"),
    Total_Profit=("Total_Profit", "sum"),
    Average_CVI=("CVI", "mean")
).reset_index()

print("========== CUSTOMER VALUE INDEX ==========")
display(cvi_summary.round(2))

print("\nTop 10 High-Value Customers:")
display(
    cvi[cvi["Customer_Value_Category"] == "High Value"]
    .sort_values("CVI", ascending=False)
    [["Customer Id", "Total_Revenue", "Total_Profit",
      "Total_Records", "CVI"]]
    .head(10).round(2)
)
```

```python
# Create discount bands using Order Item Discount Rate

bins = [-0.000001, 0, 0.05, 0.10, 0.15, 0.20, 0.25]

labels = [
    "0% Discount",
    "Above 0%–5%",
    "Above 5%–10%",
    "Above 10%–15%",
    "Above 15%–20%",
    "Above 20%–25%"
]

df["Discount_Band"] = pd.cut(
    df["Order Item Discount Rate"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

# Analyze profitability for each discount band

discount_analysis = df.groupby(
    "Discount_Band", observed=True
).agg(
    Total_Records=("Sales", "count"),
    Total_Revenue=("Sales", "sum"),
    Total_Profit=("Benefit per order", "sum"),
    Average_Profit=("Benefit per order", "mean"),
    Average_Discount=("Order Item Discount", "mean"),
    Average_Discount_Rate=("Order Item Discount Rate", "mean"),
    Loss_Records=("Benefit per order", lambda x: (x < 0).sum())
).reset_index()

# Calculate profit margin and loss percentage

discount_analysis["Profit_Margin"] = (
    discount_analysis["Total_Profit"] /
    discount_analysis["Total_Revenue"]
) * 100

discount_analysis["Loss_Record_Percentage"] = (
    discount_analysis["Loss_Records"] /
    discount_analysis["Total_Records"]
) * 100

display(discount_analysis.round(2))
```

```python
# Group category profitability by discount band

category_discount = df.groupby(
    ["Category Name", "Discount_Band"],
    observed=True
).agg(
    Total_Records=("Sales", "count"),
    Total_Revenue=("Sales", "sum"),
    Total_Profit=("Benefit per order", "sum")
).reset_index()

# Calculate profit margin
category_discount["Profit_Margin"] = (
    category_discount["Total_Profit"] /
    category_discount["Total_Revenue"]
) * 100

# Create a margin comparison table
margin_pivot = category_discount.pivot(
    index="Category Name",
    columns="Discount_Band",
    values="Profit_Margin"
)

# Get transaction counts for both extreme discount bands
count_pivot = category_discount.pivot(
    index="Category Name",
    columns="Discount_Band",
    values="Total_Records"
)

no_discount = "0% Discount"
high_discount = "Above 20%–25%"

# Keep categories with at least 20 records in both bands
valid_categories = (
    (count_pivot[no_discount] >= 20) &
    (count_pivot[high_discount] >= 20)
)

comparison = margin_pivot.loc[valid_categories, [
    no_discount, high_discount
]].copy()

comparison["Margin_Change"] = (
    comparison[high_discount] - comparison[no_discount]
)

comparison["No_Discount_Records"] = count_pivot.loc[
    valid_categories, no_discount
]

comparison["High_Discount_Records"] = count_pivot.loc[
    valid_categories, high_discount
]

comparison = comparison.sort_values("Margin_Change")

display(comparison.round(2))
```

```python
# Calculate category-level profitability and loss rates

category_discount = df.groupby(
    ["Category Name", "Discount_Band"],
    observed=True
).agg(
    Total_Records=("Sales", "count"),
    Total_Revenue=("Sales", "sum"),
    Total_Profit=("Benefit per order", "sum"),
    Loss_Records=("Benefit per order", lambda x: (x < 0).sum())
).reset_index()

category_discount["Profit_Margin"] = (
    category_discount["Total_Profit"] /
    category_discount["Total_Revenue"]
) * 100

category_discount["Loss_Rate"] = (
    category_discount["Loss_Records"] /
    category_discount["Total_Records"]
) * 100

# Create comparison tables
no_discount = "0% Discount"
high_discount = "Above 20%–25%"

margin_pivot = category_discount.pivot(
    index="Category Name",
    columns="Discount_Band",
    values="Profit_Margin"
)

count_pivot = category_discount.pivot(
    index="Category Name",
    columns="Discount_Band",
    values="Total_Records"
)

loss_pivot = category_discount.pivot(
    index="Category Name",
    columns="Discount_Band",
    values="Loss_Rate"
)

# Compare only categories with at least 100 records in both bands
valid = (
    (count_pivot[no_discount] >= 100) &
    (count_pivot[high_discount] >= 100)
)

reliable_comparison = pd.DataFrame({
    "No_Discount_Margin": margin_pivot.loc[valid, no_discount],
    "High_Discount_Margin": margin_pivot.loc[valid, high_discount],
    "Margin_Change": (
        margin_pivot.loc[valid, high_discount] -
        margin_pivot.loc[valid, no_discount]
    ),
    "No_Discount_Records": count_pivot.loc[valid, no_discount],
    "High_Discount_Records": count_pivot.loc[valid, high_discount],
    "No_Discount_Loss_Rate": loss_pivot.loc[valid, no_discount],
    "High_Discount_Loss_Rate": loss_pivot.loc[valid, high_discount]
})

reliable_comparison = reliable_comparison.sort_values(
    "Margin_Change"
)

display(reliable_comparison.round(2))
```

```python
# Compare possible profit-ratio formulas

profit = df["Benefit per order"]
dataset_ratio = df["Order Item Profit Ratio"]

candidate_columns = [
    "Sales",
    "Sales per customer",
    "Order Item Total",
    "Product Price"
]

results = []

for col in candidate_columns:
    calculated = profit / df[col]

    results.append({
        "Denominator": col,
        "Average_Ratio": calculated.mean(),
        "Median_Ratio": calculated.median(),
        "Correlation_With_Dataset": calculated.corr(dataset_ratio),
        "Mean_Absolute_Difference": (
            dataset_ratio - calculated
        ).abs().mean(),
        "Maximum_Absolute_Difference": (
            dataset_ratio - calculated
        ).abs().max()
    })

ratio_comparison = pd.DataFrame(results)

display(ratio_comparison.round(4))

# Inspect a few individual records
sample = df[[
    "Sales",
    "Sales per customer",
    "Order Item Total",
    "Product Price",
    "Benefit per order",
    "Order Item Profit Ratio"
]].head(10).copy()

sample["Calculated_Ratio"] = (
    sample["Benefit per order"] / sample["Sales"]
)

display(sample.round(4))
```

```python
# CUSTOMER VALUE INDEX (CVI)

cvi = customer_analysis.copy()

# Calculate individual scores using percentile ranking
cvi["Profit_Score"] = cvi["Total_Profit"].rank(pct=True) * 100

cvi["Revenue_Score"] = cvi["Net_Revenue"].rank(pct=True) * 100

cvi["Frequency_Score"] = cvi["Total_Records"].rank(pct=True) * 100

# Calculate weighted Customer Value Index
cvi["CVI"] = (
    0.50 * cvi["Profit_Score"] +
    0.30 * cvi["Revenue_Score"] +
    0.20 * cvi["Frequency_Score"]
)

# Classify customers
cvi["Customer_Value_Category"] = np.where(
    cvi["Total_Profit"] < 0,
    "Loss-making",
    np.where(
        cvi["CVI"] >= 80,
        "High Value",
        np.where(cvi["CVI"] >= 40, "Medium Value", "Low Value")
    )
)

# Summary by customer value category
cvi_summary = cvi.groupby("Customer_Value_Category").agg(
    Total_Customers=("Customer Id", "count"),
    Net_Revenue=("Net_Revenue", "sum"),
    Total_Profit=("Total_Profit", "sum"),
    Average_CVI=("CVI", "mean")
).reset_index()

cvi_summary["Profit_Contribution"] = (
    cvi_summary["Total_Profit"] / total_profit
) * 100

print("========== CUSTOMER VALUE CATEGORY SUMMARY ==========")
display(cvi_summary.round(2))

# Top 10 high-value customers
print("\n========== TOP 10 HIGH-VALUE CUSTOMERS ==========")

top_high_value = cvi[
    cvi["Customer_Value_Category"] == "High Value"
].sort_values(by="CVI", ascending=False)

display(
    top_high_value[
        [
            "Customer Id",
            "Net_Revenue",
            "Total_Profit",
            "Total_Records",
            "CVI",
            "Customer_Value_Category"
        ]
    ].head(10).round(2)
)
```
